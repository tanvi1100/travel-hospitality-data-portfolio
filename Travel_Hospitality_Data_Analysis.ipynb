# Travel & Hospitality Data Analysis

## Week 1: Hotel Booking Analysis

### Objective
Analyze hotel booking data to understand pricing trends, destination demand, and customer behavior using Python and Pandas.

```python
import pandas as pd
```

```python
data = [
    (1, 101, 'Luxury Stay Hotel', 'New York', 450, 4.8, '2026-01-01'),
    (2, 102, 'Ocean View Resort', 'Miami', 320, 4.5, '2026-01-02'),
    (3, 103, 'City Comfort Inn', 'Chicago', 180, 4.0, '2026-01-03'),
    (4, 104, 'Royal Palace Hotel', 'New York', 600, 4.9, '2026-01-04'),
    (5, 105, 'Beachside Escape', 'Los Angeles', 520, 4.7, '2026-01-05'),
    (6, 106, 'Urban Stay', 'Chicago', 200, 3.9, '2026-01-06'),
    (7, 107, 'Grand Luxe Hotel', 'Miami', 700, 4.9, '2026-01-07'),
    (8, 108, 'Budget Comfort', 'Dallas', 120, 3.8, '2026-01-08'),
    (9, 109, 'Skyline Suites', 'New York', 800, 4.9, '2026-01-09'),
    (10, 110, 'Sunset Resort', 'Los Angeles', 400, 4.6, '2026-01-10')
]

columns = [
    "booking_id",
    "user_id",
    "hotel_name",
    "city",
    "price",
    "rating",
    "booking_date"
]

df = pd.DataFrame(data, columns=columns)

df.head()
```

First **Analysis**

```python
print("Total Bookings:", len(df))
print("Average Booking Price:", round(df["price"].mean(), 2))
```

Popular **Cities**

```python
df.groupby("city").size().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

city_prices = df.groupby("city")["price"].mean().sort_values()

city_prices.plot(kind="bar")

plt.title("Average Hotel Price by City")
plt.xlabel("City")
plt.ylabel("Average Price")
plt.show()
```

## Business Insights (Week 1)

- New York has the highest average hotel prices, indicating strong luxury and premium travel demand.
- Dallas has the lowest average prices, representing a budget travel market.
- Miami and Los Angeles show mid-to-high pricing, suggesting strong leisure and vacation demand.
- Chicago reflects a balanced market with moderate pricing and steady demand.

## Key Takeaway
The dataset shows clear geographic segmentation in travel demand, which is essential for pricing strategy and hotel expansion decisions.
